# Lista 3 – Variáveis e Tuplas
**Objetivo:** Trabalhar com criação, manipulação, busca e concatenação de tuplas em Python usando símbolos de aminoácidos.

### Exercício Único: Manipulação de Tuplas de Aminoácidos

```python
# Construindo a tupla inicial com os aminoácidos
aminoacidos = ('A', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'K', 'L', 'M', 'N', 'P', 'Q', 'R', 'S', 'T', 'V', 'W', 'Y')

# a) Conte e imprima o número de elementos contidos na tupla criada.
print("a) Número de elementos na primeira tupla:", len(aminoacidos))

# b) Verifique e imprima se o aminoácido Serina (S) pertence à tupla.
print("b) Serina (S) pertence à tupla?", 'S' in aminoacidos)

# c) Crie uma segunda tupla contendo Prolina (P), Glicina (G), Asparagina (N), Tirosina (Y), Valina (V) e Triptofano (W).
# Obs: O caractere da Valina estava vazio no PDF original, mas sabemos que é 'V'.
tupla2 = ('P', 'G', 'N', 'Y', 'V', 'W')
print("c) Segunda tupla criada:", tupla2)

# d) Some as tuplas construídas e imprima o resultado.
tupla_somada = aminoacidos + tupla2
print("d) Tuplas somadas:\n", tupla_somada)

# e) Conte a ocorrência dos elementos Glicina (G), Asparagina (N) e Cisteína (C).
# Contando na tupla resultante da soma, onde alguns elementos aparecem mais de uma vez.
cont_g = tupla_somada.count('G')
cont_n = tupla_somada.count('N')
cont_c = tupla_somada.count('C')
print(f"e) Ocorrências -> G: {cont_g}, N: {cont_n}, C: {cont_c}")

# f) Retorne a posição do primeiro elemento Asparagina (N).
pos_n = tupla_somada.index('N')
print("f) Posição do primeiro 'N':", pos_n)

# g) Retorne os 5 últimos elementos da tupla.
ultimos_5 = tupla_somada[-5:]
print("g) Últimos 5 elementos:", ultimos_5)
```
