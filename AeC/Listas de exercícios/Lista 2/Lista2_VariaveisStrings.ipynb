### Exercício 1: Manipulação de Sequência de Aminoácidos (TNF alfa)

```python
# Definindo a sequência do TNF alfa
tnf_alfa = "VRSSSRTPSDKPVAHVVANPQAEGQLQWLNRRANALLANGVELRDNQLVVPSEGLYLIYSQVLFKGQGCPSTHVLLTHTISRIAVSYQTKVNLLSAIKSPCQRETPEGAEAKPWYEPIYLGGVFQLEKGDRLSAEINRPDYLLFAESGQVYFGIIAL"

# a) Retorne o tamanho da sequência.
tamanho = len(tnf_alfa)
print("a) Tamanho da sequência:", tamanho)

# b) Realize a contagem da ocorrência de uma sequência de leucinas (LL).
contagem_ll = tnf_alfa.count("LL")
print("b) Ocorrências de LL:", contagem_ll)

# c) Encontre na sequência as posições ocupadas por duas glicinas (GG) e duas argininas (RR).
posicao_gg = tnf_alfa.find("GG")
posicao_rr = tnf_alfa.find("RR")
print("c) Posição de GG:", posicao_gg, "| Posição de RR:", posicao_rr)

# d) Retorne os 100 primeiros aminoácidos da sequência.
primeiros_100 = tnf_alfa[:100]
print("d) 100 primeiros aminoácidos:", primeiros_100)

# e) Substitua o trecho da sequência com a ocorrência de 3 serinas e 1 arginina (SSSR) por alaninas.
# Como o trecho original tem 4 letras, usaremos 4 alaninas (AAAA).
seq_substituida = tnf_alfa.replace("SSSR", "AAAA")
print("e) Sequência com substituição:\n", seq_substituida)

# f) Quebre a sequência no local onde a substituição foi realizada.
seq_quebrada = seq_substituida.split("AAAA")
print("f) Sequência quebrada em lista:", seq_quebrada)
```

### Exercício 2: Manipulação de Texto

```python
texto = "As proteínas são cadeias polipeptídicas formadas pela ligação peptidica entre resíduos de aminoácidos. Existem 20 tipos de aminoácidos comumente encontrados nos seres vivos. A esses aminoácidos, foram atribuídas abreviações de 3 letras e símbolos de 1 letra. As abreviações de 3 letras são bastante evidentes consistindo nas três primeiras letras do se nome."

# a) Passe todo o texto para letras maiúsculas.
print("a) Maiúsculas:\n", texto.upper(), "\n")

# b) Passe todo o texto para letras minúsculas.
print("b) Minúsculas:\n", texto.lower(), "\n")

# c) Passe cada primeira letra de palavra em maiúsculo.
print("c) Iniciais maiúsculas:\n", texto.title(), "\n")

# d) Transforme as letras maiúsculas em minúsculas e vice-versa.
print("d) Invertendo maiúsculas/minúsculas:\n", texto.swapcase())
```

### Exercício 3: Manipulação de sequência de insulina

```python
insulin_signal = "MALWMRLLPLLALLALWGPDPAAA"

# a) Retorne o tamanho da sequência apresentada.
print("a) Tamanho da sequência:", len(insulin_signal))

# b) Quebre a sequência no trecho \"LLALLALWG\".
partes_insulina = insulin_signal.split("LLALLALWG")
print("b) Sequência quebrada:", partes_insulina)

# c) Concatene as sequências resultantes obtendo a seguinte sequência final MALWMRLLPPDPAAA.
insulina_concatenada = "".join(partes_insulina)
print("c) Sequência concatenada:", insulina_concatenada)

# d) Substitua o trecho \"DPAAA\" por \"LLALL\".
insulina_final = insulina_concatenada.replace("DPAAA", "LLALL")
print("d) Nova sequência com substituição:", insulina_final)
```

### Exercício 4: Transcrição de DNA para RNA

```python
dna = "GATGGAACTTGACGTAAACCTATATT"

# O RNA substitui a base Timina (T) do DNA por Uracila (U)
rna = dna.replace("T", "U")

print("Sequência de DNA: ", dna)
print("Sequência de RNA: ", rna)
```
