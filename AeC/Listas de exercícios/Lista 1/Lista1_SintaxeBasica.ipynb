# Lista 1 – Sintaxe Básica
**Objetivo:** Digitar, executar e entender os trechos de código em Python focados em sintaxe básica.

### Exercício 1: Atribuição e impressão de string

```python
# Atribui a palavra "python" à variável 'linguagem' e a exibe na tela
linguagem = "python"
print(linguagem)
```

### Exercício 2: Operações matemáticas básicas

```python
# Realiza e imprime operações de soma, multiplicação e subtração entre duas variáveis
a = 1500
b = 335
print(a + b)
print(a * b)
print(a - b)
```

### Exercício 3: Concatenação de strings

```python
# Junta (concatena) três strings diferentes e imprime o resultado final
a = "Marie"
b = " Curie " 
c = "!!!"
print(a + b + c)
```

### Exercício 4: Expressões booleanas (Verdadeiro)

```python
# Verifica se a soma de 2 + 3 é igual a 5. O resultado esperado é 'True'
print(2 + 3 == 5)
```

### Exercício 5: Expressões booleanas (Falso)

```python
# Fazer uma atribuição dentro da expressão de Print gera um erro de Syntax
print(5 + 9 = 13)
```

### Exercício 6: Variáveis do tipo Float e comentários

```python
# Atribui um valor decimal (float) à variável pi e a imprime
pi = 3.1415 # Valor do número pi
print(pi)
```
