**Objetivo:** Praticar operações de teoria dos conjuntos em Python (interseção, diferença, união/atualização e verificação de subconjuntos/disjuntos) usando dados de RMSD de docking molecular e listas numéricas genéricas.

### Exercício 1: Conjuntos de RMSD (Docking Molecular)

```python
# Inicializando os 3 conjuntos com base na tabela do experimento
ferramenta1 = {1.9, 1.8, 5.7, 1.6, 5.8, 1.7, 9.6, 5.9, 9.5, 6.5, 6.2, 1.1, 4.4, 3.5, 2.9, 4.7, 4.6, 5.2, 5.3}
ferramenta2 = {4.7, 3.6, 6.2, 7.1, 7, 5.6, 5.7, 3.4, 3.3, 2.1, 3.8, 3.9, 5, 5.1, 1.9, 9.5, 1.0, 1.3, 5.4}
ferramenta3 = {2.2, 3.3, 5.1, 3, 3.7, 9.1, 8.8, 8.5, 2, 4.1, 6.1, 4.9, 1.1, 0.5, 0.8, 3.2, 6.9, 9.3, 9.5}

# a) Verifique as diferenças entre os conjuntos (1,2), (1,3) e (2,3)
print("a) Diferença entre Ferramenta 1 e 2 (1 - 2):", ferramenta1.difference(ferramenta2))
print("   Diferença entre Ferramenta 1 e 3 (1 - 3):", ferramenta1.difference(ferramenta3))
print("   Diferença entre Ferramenta 2 e 3 (2 - 3):", ferramenta2.difference(ferramenta3))

# b) Retorne as intersecções entre (1,2), (1,3) e (2,3)
print("\nb) Interseção entre 1 e 2:", ferramenta1.intersection(ferramenta2))
print("   Interseção entre 1 e 3:", ferramenta1.intersection(ferramenta3))
print("   Interseção entre 2 e 3:", ferramenta2.intersection(ferramenta3))

# c) Insira todos os elementos do conjunto 2 e 3 no conjunto 1
ferramenta1.update(ferramenta2, ferramenta3)
print("\nc) Conjunto 1 após receber os elementos de 2 e 3:\n", ferramenta1)

# d) Retorne o tamanho do conjunto formado pela tarefa (c)
print("\nd) Tamanho final do conjunto 1 atualizado:", len(ferramenta1))
```

### Exercício 2: Operações Genéricas com Conjuntos

```python
# Inicializando os conjuntos
A = {3, 6, 9, 12, 15, 18, 21, 24, 28, 27}
B = {2, 6, 8, 10, 14, 16, 18, 20, 22, 24}
C = {2, 6, 10, 18, 20}
D = {1, 30, 5, 11, 17, 16, 22, 26}

# a) Verifique a interseção e diferença entre o conjunto A e B
print("a) Interseção entre A e B:", A.intersection(B))
print("   Diferença entre A e B (A - B):", A.difference(B))

# b) Verifique se o conjunto A e B são disjuntos ao conjunto D
# Conjuntos disjuntos não possuem elementos em comum.
print("\nb) Conjunto A é disjunto de D?", A.isdisjoint(D))
print("   Conjunto B é disjunto de D?", B.isdisjoint(D))

# c) Retorne se o conjunto C é subconjunto de A e B
print("\nc) O conjunto C é subconjunto de A?", C.issubset(A))
print("   O conjunto C é subconjunto de B?", C.issubset(B))

# d) Acrescente os elementos 18, 23, 25, 63 no conjunto D
D.update({18, 23, 25, 63})
print("\nd) Conjunto D após acréscimo dos novos elementos:\n", D)
```
