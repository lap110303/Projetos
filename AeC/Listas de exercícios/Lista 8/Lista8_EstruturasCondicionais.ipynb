# Lista 8 – Estruturas Condicionais
**Objetivo:** Utilizar estruturas condicionais (`if`, `elif`, `else`) para aplicar filtros e lógicas de comparação em sequências de aminoácidos.

```python
# Definição das sequências
seq_a = "LRSSSQNSSDKPVAHVVANHQVEEQLEWLSQRANALLANGMDLKDNQLVVPADGLYLVYSQVLFKGQGCPDYVLLTHTVSLRSSSDK"
seq_b = "KPAAHLIGDPSKQNSLLWRANTDRAFLQDGFSLSNNSLLVPTSGIYFVYSQVVFSGKAYSPKATSSPLYLAHEVQLFSS"
seq_c = "CPQGKYIHPQNNSICCTKCHKGTYLYNDCPGPGQDTDCRECESGSFTASENHLRHCLSCSKCRKEMGQVEISSCTVDRDTVCGCR"

# Criação de um dicionário para facilitar as iterações e manter o nome das sequências
sequencias = {
    "Sequência A": seq_a,
    "Sequência B": seq_b,
    "Sequência C": seq_c
}
```

### Exercício 1: Sequências com 80 ou mais aminoácidos

```python
print("1. Sequências com 80 ou mais aminoácidos:")
for nome, seq in sequencias.items():
    if len(seq) >= 80:
        print(f"{nome} (Tamanho: {len(seq)})")
```

### Exercício 2: Sequências com tamanho maior que a média

```python
media_tamanho = sum(len(seq) for seq in sequencias.values()) / len(sequencias)
print(f"2. Média de tamanho: {media_tamanho:.2f}")
print("Sequências maiores que a média:")

for nome, seq in sequencias.items():
    if len(seq) > media_tamanho:
        print(f"{nome} (Tamanho: {len(seq)})")
```

### Exercício 3: Sequências com pelo menos uma Histidina (H) e nenhuma Prolina (P)

```python
print("3. Sequências com H (Histidina) e sem P (Prolina):")
for nome, seq in sequencias.items():
    if 'H' in seq and 'P' not in seq:
        print(f"{nome} atende a condição.")
else:
    # Verifica se a saída ficou vazia (caso nenhuma atenda)
    if not any('H' in seq and 'P' not in seq for seq in sequencias.values()):
        print("Nenhuma das sequências atende a este critério.")
```

### Exercício 4: Maior dentre as três sequências

```python
maior_nome = ""
maior_tamanho = 0

for nome, seq in sequencias.items():
    if len(seq) > maior_tamanho:
        maior_nome = nome
        maior_tamanho = len(seq)

print(f"4. A maior sequência é a {maior_nome} com {maior_tamanho} aminoácidos.")
```

### Exercício 5: Sequências em ordem crescente de tamanho

```python
print("5. Sequências ordenadas por tamanho crescente:")
# Utiliza sorted com uma função lambda para ordenar pelo tamanho da sequência (x[1])
sequencias_ordenadas = sorted(sequencias.items(), key=lambda x: len(x[1]))

for nome, seq in sequencias_ordenadas:
    print(f"{nome}: {len(seq)} aminoácidos")
```
