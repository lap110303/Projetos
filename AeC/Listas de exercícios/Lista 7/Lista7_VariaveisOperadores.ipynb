**Objetivo:** Aplicar operadores matemáticos em problemas de bioinformática, como o cálculo de RMSD (Root Mean Square Deviation) de coordenadas atômicas e o conteúdo GC de sequências genéticas.

### Exercício 1: Cálculo de RMSD (Glicina)

```python
import math

# Coordenadas atômicas (X, Y, Z) para Glicina 1 [N, CA, C, O]
x1 = [108.304, 108.477, 109.907, 110.821]
y1 = [100.827, 100.389, 100.555, 100.799]
z1 = [67.992, 69.362, 69.817, 69.027]

# Coordenadas atômicas (X, Y, Z) para Glicina 2 [N, CA, C, O]
x2 = [107.670, 108.477, 109.513, 110.667]
y2 = [101.359, 100.389, 101.011, 100.572]
z2 = [70.074, 69.362, 68.450, 68.425]

N = 4 # Número de átomos
soma_diferencas_quadrado = 0

# Aplicando o somatório para calcular a distância euclidiana ao quadrado de cada átomo
for i in range(N):
    soma_diferencas_quadrado += (x1[i] - x2[i])**2 + (y1[i] - y2[i])**2 + (z1[i] - z2[i])**2

# Finalizando a fórmula do RMSD (Raiz quadrada da média)
rmsd = math.sqrt((1/N) * soma_diferencas_quadrado)

print(f"O valor do RMSD calculado entre os resíduos de Glicina é: {rmsd:.3f}")
```

### Exercício 2: Cálculo do percentual de conteúdo GC

```python
seq_a = "ATGATCTCGTAATTAACCGGAATTTTGGGCC"
seq_b = "GGCCTTAAGTTTAACCCGGAATTTAAAGGCCCCAAA"

def calcular_conteudo_gc(sequencia):
    # Conta ocorrências de G e C
    total_g = sequencia.count('G')
    total_c = sequencia.count('C')
    
    # Calcula o percentual em relação ao tamanho total
    percentual = ((total_g + total_c) / len(sequencia)) * 100
    return percentual

gc_seq_a = calcular_conteudo_gc(seq_a)
gc_seq_b = calcular_conteudo_gc(seq_b)

print(f"Conteúdo GC da Sequência A: {gc_seq_a:.2f}%")
print(f"Conteúdo GC da Sequência B: {gc_seq_b:.2f}%")
```
