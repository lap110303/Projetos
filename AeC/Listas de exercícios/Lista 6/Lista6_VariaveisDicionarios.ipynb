# Lista 6 – Variáveis e Dicionários
**Objetivo:** Trabalhar com a estrutura de dados Dicionário em Python, utilizando identificadores do Protein Data Bank (PDB) e suas respectivas contagens de aminoácidos.

### Exercício Único: Manipulação de Dicionários PDB

```python
# a) Construa um dicionário onde os ids PDB sejam as chaves e o número de resíduos sejam o valor.
pdb_dict = {
    "1A8M": 471,
    "1TNR": 283,
    "2AZ5": 592,
    "1TNF": 471,
    "2TNF": 468,
    "2TUN": 942,
    "4TSV": 150,
    "5TSW": 900,
    "2E7A": 471,
    "6RMJ": 489
}
print("a) Dicionário criado:\n", pdb_dict)

# b) Imprima os valores contidos nos ids 2TNF e 2E7A.
print("\nb) Valor para 2TNF:", pdb_dict["2TNF"])
print("   Valor para 2E7A:", pdb_dict["2E7A"])

# c) Verifique e imprima o tamanho do dicionário construído.
print("\nc) Tamanho do dicionário:", len(pdb_dict))

# d) Obtenha e imprima a lista de todas as chaves do dicionário.
chaves = list(pdb_dict.keys())
print("\nd) Lista de chaves:\n", chaves)

# e) Obtenha e imprima a lista de todos os valores do dicionário.
valores = list(pdb_dict.values())
print("\ne) Lista de valores:\n", valores)

# f) Obtenha e imprima uma lista de tuplas com todos os pares chave-valor.
pares = list(pdb_dict.items())
print("\nf) Lista de tuplas (chave, valor):\n", pares)
```
